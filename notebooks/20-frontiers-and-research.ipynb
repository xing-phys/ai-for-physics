# From accurate predictions to scientific progress {#discovery-frontiers}

The methods in this book share a recurring structure. A physical question defines observables; a representation restricts possible answers; an optimization or inference procedure selects an answer; and independent evidence establishes its usefulness. Scientific progress occurs when this chain permits a calculation, measurement, prediction, or explanation that was previously inaccessible.

This chapter maps research directions through that structure. The literature snapshot is current to **21 October 2026**; it is selective rather than exhaustive. Recent results are presented as evidence for particular tasks, not permanent rankings. The distinction between a mathematical guarantee, a finite numerical observation, and a physical discovery is especially important at the frontier, where impressive benchmarks can coexist with unresolved assumptions.

## Quantum many-body physics: what has actually converged?

Neural quantum states address an old obstruction: a generic many-body wavefunction requires exponentially many amplitudes. A compact ansatz is useful if it represents the relevant states, can be optimized, and permits reliable estimates of target observables. These are three different challenges. A more expressive network does not ensure a successful optimizer; an accurate energy does not establish every correlation function.

Consider a normalized trial state expanded in exact energy eigenstates, $|\psi\rangle=\sum_nc_n|n\rangle$, with a unique ground state $|0\rangle$ and gap $\Delta=E_1-E_0>0$. Then

$$
E_\psi-E_0=\sum_{n>0}|c_n|^2(E_n-E_0)
\geq\Delta\sum_{n>0}|c_n|^2
=\Delta(1-|c_0|^2).
$$

Thus the ground-state fidelity obeys $|c_0|^2\geq1-(E_\psi-E_0)/\Delta$. An energy error is informative only relative to the relevant gap. If the gap shrinks with system size, a small energy error per site can coexist with substantial state error. Degenerate ground spaces require replacing ground-state fidelity by total weight in that space.

A zero energy variance also needs interpretation. Since $\operatorname{Var}(H)=\|(H-E_\psi)|\psi\rangle\|^2$, zero variance means an exact eigenstate, not necessarily the ground state. Jointly inspect energy, variance, symmetries, correlations, and comparisons with other methods. For dynamics, check conserved charges and accumulated phase errors as well as short-time observables.

A significant computational direction is to improve the geometry and scaling of optimization. Minimum-step stochastic reconfiguration enabled larger neural ansätze and improved variational calculations in frustrated spin models in a 2024 study [@q-chen2024]. Its conclusions about spin-liquid gaps are numerical evidence involving finite-size extrapolation, not a theorem settling all frustrated magnets. Another direction conditions a shared wavefunction representation on Hamiltonian couplings. A 2025 foundation neural-state study investigated transfer across Hamiltonians and fine-tuning [@d-rende2025]. The unresolved scientific question is which physical families share reusable structure, particularly across changes of phase, geometry, sign structure, or entanglement.

The same caution applies to quantum dynamics and open systems. A model may reproduce local observables while missing rare trajectories, long-time transport, or an exponentially small decay rate. Useful research projects connect the ansatz to a controlled limit: exact diagonalization at small size, tensor networks when entanglement is manageable, sign-free Monte Carlo where available, perturbative expansions, or conservation identities. Agreement among methods with different error mechanisms is stronger evidence than agreement among similar neural architectures.

## Atomistic physics: forces, stability, and synthesis are separate tests

Interatomic potentials aim to replace repeated electronic-structure calculations with a learned energy $E_\theta(R,Z)$ for atomic positions $R$ and species $Z$. Differentiation gives forces $F=-\nabla_RE_\theta$. Enforcing rotation, translation, and permutation symmetry reduces the burden on data, but energy-conserving forces can still be quantitatively wrong.

Near an equilibrium structure $R_0$, expand

$$
E(R_0+u)=E(R_0)+\tfrac12u^TKu+O(\|u\|^3).
$$

The Hessian $K$ controls small vibrations. For masses collected in diagonal matrix $M$, normal modes satisfy $Ke=\omega^2Me$. A potential can have a small average energy error while misestimating curvature, phonons, or activation barriers. A 2025 study documented systematic softening in several universal potentials on its selected out-of-distribution tests [@d-deng2025]. This is a benchmark observation about models and datasets, not a universal impossibility result. It motivates training data that cover displaced, defective, and high-energy configurations as well as equilibrium structures.

Material stability requires another calculation. A compound of composition $c$ is stable against a collection of competing phases if its energy is no higher than any composition-preserving mixture. If mixture fractions $w_i\geq0$ satisfy $\sum_iw_i=1$ and $\sum_iw_ic_i=c$, the reference energy is the minimum of $\sum_iw_iE_i$. The difference above this minimum is the energy above the convex hull. Its uncertainty includes errors in the target and competing phases. A low predicted formation energy alone does not establish stability; a missing competing phase can change the hull.

Generative models extend screening by proposing structures rather than only scoring an existing database. MatterGen demonstrated property-conditioned generation with computational evaluation and a targeted experimental example [@d-zeni2025]. This establishes a useful route to candidates. Broad claims of synthesizability require kinetic pathways, finite-temperature effects, disorder, precursor availability, and experimental characterization. A generated structure that is absent from one database is not automatically a new physical material. Novelty needs structure-aware matching and a dated search of relevant sources.

Promising research therefore connects a hierarchy: candidate generation, accurate relaxation, stability and property uncertainty, synthesis planning, and independent characterization. Every stage can reject candidates. Reporting the complete funnel, including rejected proposals and computational costs, makes a discovery claim easier to assess than reporting only the final successful examples.

## High-energy physics: learning without manufacturing significance

Collider detectors produce variable-length sets of tracks and energy deposits. Graphs and attention-based models can exploit this structure for reconstruction, tagging, simulation, unfolding, and inference. A learned representation shared across tasks may reduce repeated training; OmniLearn investigated this approach for jet tasks and transfer between selected datasets [@d-mikuni2024]. Transfer is an empirical property of a model under tested domain changes, not a consequence of calling it a foundation model.

A physical discovery needs a statistical comparison with plausible alternatives, including detector and theoretical uncertainties. Suppose an anomaly score identifies a region with a small background probability $p$. If $m$ independent regions were searched, the probability of at least one such fluctuation is $1-(1-p)^m$, approximately $mp$ when $mp\ll1$. Real searches are correlated; the effective trials factor usually requires dedicated simulations. Choosing the anomaly score, cuts, and mass window after seeing the signal can further enlarge the selection process.

An anomaly detector can also learn a simulator deficiency. Validate score distributions in control regions, propagate nuisance parameters, test alternative generators, and design signal-injection studies. If the method uses a sideband to predict the background in a signal region, state the smoothness or independence assumptions required by that extrapolation. A classifier with high test accuracy may still be unsuitable for precision inference if it is sensitive to poorly controlled detector effects.

Generative detector simulation has a related challenge: average image similarity can hide rare tails that drive a discovery analysis. Compare physically meaningful joint observables, response conditional on incident particles, conservation constraints, extreme deposits, and downstream statistical conclusions. The final standard is whether the analysis retains correct sensitivity and false-positive behavior, not whether simulated events look plausible.

## Astrophysics: amortized inference meets selection effects

Neural density estimators can accelerate repeated gravitational-wave or cosmological parameter inference. The conceptual advantage is amortization: training is expensive, but many later observations can reuse the same simulator-informed model. Accounting for training cost, retraining frequency, detector drift, and posterior validation is essential when reporting the practical speedup.

Population inference illustrates why fast individual posteriors are only one component. Let $\lambda$ describe a population distribution $p(\theta\mid\lambda)$ and let detection probability be $P_{\mathrm{det}}(\theta)$. The distribution of parameters among detected events is

$$
p(\theta\mid\lambda,\mathrm{det})
=\frac{P_{\mathrm{det}}(\theta)p(\theta\mid\lambda)}{\alpha(\lambda)},
\qquad
\alpha(\lambda)=\int P_{\mathrm{det}}(\theta)p(\theta\mid\lambda)d\theta.
$$

This follows by conditioning the joint probability of parameters and detection. Ignoring the denominator favors populations that are intrinsically easier to detect. The full likelihood also depends on whether the number of events is conditioned on or modeled as a counting process. A model that estimates each event accurately can still yield biased cosmology if selection or population assumptions are wrong.

For imaging, lensing, and cosmological fields, assess which scales and features are constrained by observations and which are supplied by the learned prior. Held-out simulations with different baryonic prescriptions, instrument responses, or noise correlations test model robustness. Applying the same simulator for training and validation primarily tests computational consistency; observational comparisons and alternative physical models probe additional questions.

## Plasma physics: prediction is useful when it changes an action

Plasmas combine nonlinear dynamics, partial observations, rapidly growing instabilities, and expensive experiments. Learned models can predict instability risk, reconstruct equilibrium, accelerate transport calculations, and guide control. A 2024 DIII-D demonstration used reinforcement learning to reduce tearing-instability risk within its experimental conditions [@d-seo2024]. The practical achievement connects prediction to a timely intervention.

The open issue is robustness when a policy changes the distribution of states it encounters. Training data may contain mostly routine shots, while a controller deliberately visits a new high-performance regime. That is a feedback-induced distribution shift. Evaluate the dynamics model on action sequences resembling the proposed policy, quantify uncertainty in poorly sampled regimes, and test recovery after unexpected perturbations. A policy that optimizes a surrogate instability score may exploit a score error unless independent diagnostics and operational constraints check it.

Control also creates a scientific opportunity: deliberately varied, recorded actions can reveal response functions more informatively than passive data. Experiments designed jointly for performance and information could improve future models. This is a research direction, not a guarantee; exploration has real time and hardware costs and must respect the facility's operating envelope.

## Weather and climate: a distribution is part of the prediction

Weather asks about the evolving state conditional on current observations. Climate asks about distributions under sustained conditions and forcing. A successful short-range weather model does not, by that fact alone, establish reliable climate response to altered forcing.

Generative weather forecasting targets distributions rather than only a best-guess trajectory. GenCast reported strong probabilistic forecasting results in its evaluated setting [@p-price2025]. Aurora investigated pretraining and adaptation across Earth-system forecasting tasks [@d-bodnar2025]. Both show the relevance of reusable representations and modern architectures; neither eliminates the need for independent operational evaluation, physical diagnostics, or tests under changed climate conditions.

For scalar predictions, a useful distributional score is

$$
\mathrm{CRPS}(F,y)=\mathbb E|X-y|-\tfrac12\mathbb E|X-X'|,
$$

where $X,X'$ are independent draws from forecast distribution $F$. It is the continuous ranked probability score; lower is better. The first term rewards closeness to the observation. The second offsets the tendency to collapse all predictions to one point. For an empirical ensemble, replace expectations by averages over members and member pairs. Calibration asks whether predicted probabilities match frequencies; sharpness asks how concentrated forecasts are. A useful forecast needs both.

```python
import numpy as np
rng = np.random.default_rng(2001)
observed = rng.normal(size=1000)
# Both models target mean zero; one seriously understates uncertainty.
wide = rng.normal(size=(1000, 60))
narrow = 0.2*rng.normal(size=(1000, 60))

def crps(ensemble, y):
    a = np.abs(ensemble-y[:, None]).mean(axis=1)
    b = np.abs(ensemble[:, :, None]-ensemble[:, None, :]).mean(axis=(1, 2))
    return np.mean(a-0.5*b)

assert crps(wide, observed) < crps(narrow, observed)
# Small energy error need not mean high quantum-state fidelity.
gap, excited_weight = 1e-4, 0.7
energy_error = gap*excited_weight
fidelity = 1-excited_weight
assert energy_error < 1e-3 and fidelity < 0.5
print("CRPS, calibrated / narrow:", crps(wide, observed), crps(narrow, observed))
print("quantum energy error / fidelity:", energy_error, fidelity)
```

For spatial fields, marginal scores are insufficient: a model can have correct temperature distributions at every grid point but incorrect correlations, storm tracks, or energy spectra. Evaluate joint structure, rare events, and downstream quantities such as accumulated precipitation or power production. Reanalysis is itself a model-assisted reconstruction; agreement with it is not identical to agreement with independent observations.

## General experimental layouts and research agents

Choosing a continuous knob is one part of design. Another is selecting a detector geometry, optical network, or sequence of measurement operations. These discrete choices demand a representation of admissible apparatus and an evaluator that translates scientific aims into a computable objective. A September 2026 review synthesizes AI design across physics experiments [@d-klimesch2026]. Its organizing questions reinforce a general lesson: the search method is only as scientifically useful as the representation, simulator, constraints, and objective it receives.

A robust frontier program often combines flexible proposals with structured evaluation. Language models can propose code or hypotheses; equivariant networks can supply efficient physical surrogates; differentiable simulation can refine a design; exact algebra or experiments can test it. The possibility of coordinating these stages is exciting. The stronger claim that a general autonomous scientist can reliably formulate and validate arbitrary new physics remains a research aspiration, not an established result.

## How to read a frontier paper critically

First rewrite its central claim as a measurable statement. Identify the physical object, domain, observable, baseline, and resource budget. “Better quantum simulation” becomes a comparison of particular observables on specified Hamiltonians and sizes. “Discovery” becomes a verified object or relation plus a documented novelty assessment.

Then track five kinds of uncertainty separately: measurement noise, finite-sample error, numerical approximation, learned-model error, and physical-model discrepancy. Repeated seeds address only part of this list. A standard error over samples does not account for ansatz bias or simulator misspecification.

A research-ready evaluation should answer the following questions.

- Are training and test systems independent at the level of trajectories, materials, experiments, or simulation seeds, rather than merely individual correlated snapshots?
- Does the strongest relevant physical baseline receive comparable data and tuning effort?
- Do reported costs include data generation, preprocessing, optimization, and retraining as well as inference?
- Which conservation laws, symmetries, limiting cases, and independently known results were checked?
- Are extrapolation tests tied to a meaningful physical change, such as system size, coupling, geometry, forcing, or detector response?
- Can a discrepancy be attributed to an individual stage, and is the claimed scientific conclusion stable under reasonable alternatives?

A useful new project need not introduce a larger architecture. It may reveal a previously hidden failure mode, derive a reliable estimator, create a benchmark with meaningful observables, or demonstrate a controlled calculation in an inaccessible regime. Start from a scientific bottleneck, specify what evidence would resolve it, and choose the simplest method capable of producing that evidence.

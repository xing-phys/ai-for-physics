# Probability, geometry, and numerical reasoning 

## Why these tools belong together

A measurement is uncertain, a wavefunction is a vector, and an optimizer changes many parameters at once. Probability describes uncertain information; linear algebra describes directions and correlations; calculus describes how predictions respond to changes. Scientific learning needs all three. This chapter reviews them in the particular form used later, deriving the identities rather than requiring a probability or numerical-analysis course.

After this chapter you should be able to normalize and transform a probability density, calculate covariances and sampling errors, interpret singular values, use Taylor expansions to assess numerical errors, and recognize entropy and relative entropy. Chapter 3 uses these tools for learning from data; the quantum chapters use their complex-vector versions.

## Vectors, matrices, and the meaning of a norm

A column vector $x\in\mathbb R^d$ stores $d$ real components. Its Euclidean inner product is $x^Ty=\sum_a x_ay_a$ and its norm is $\|x\|=(x^Tx)^{1/2}$. For complex vectors use $x^\dagger y=\sum_a x_a^*y_a$, where $*$ means complex conjugation and $\dagger$ conjugate transpose. A matrix $A$ maps input vectors to output vectors by $(Ax)_i=\sum_jA_{ij}x_j$.

The choice of norm contains an assumption about what differences matter. Comparing displacement errors in meters with temperature errors in kelvin using one unweighted Euclidean norm is physically meaningless. Define dimensionless coordinates $\widetilde x_a=(x_a-x_{a,0})/s_a$ with specified nonzero scales $s_a$, or use a positive-definite weight matrix $W$ and $\|x\|_W^2=x^TWx$. For measurement errors of covariance $C$, $W=C^{-1}$ weights errors by their statistical precision.

A real symmetric matrix satisfies $A=A^T$ and has orthonormal eigenvectors $u_a$ with real eigenvalues $\lambda_a$:

$$
A=\sum_a\lambda_a u_au_a^T,\qquad A u_a=\lambda_a u_a.
$$

It is positive semidefinite when $v^TAv\geq0$ for every $v$, equivalent to $\lambda_a\geq0$. The complex analogue is a Hermitian matrix, $A=A^\dagger$. These facts explain why covariance matrices, quantum density matrices, and local quadratic losses have useful spectral descriptions.

### Singular values and directions we cannot see

For a rectangular matrix $X\in\mathbb R^{n\times d}$, the singular-value decomposition is $X=U\Sigma V^T$. The columns $v_a$ of $V$ are orthonormal input directions, the columns $u_a$ of $U$ are output directions, and the diagonal entries $s_a\geq0$ of $\Sigma$ obey $Xv_a=s_au_a$. To see where they come from, diagonalize $X^TX$: its nonnegative eigenvalues are $s_a^2$, and for $s_a>0$ define $u_a=Xv_a/s_a$. Then $u_a^Tu_b=v_a^TX^TXv_b/(s_as_b)=\delta_{ab}$.

A small $s_a$ means a large change in the input along $v_a$ barely changes the output. Inverting that map magnifies noise by $1/s_a$. A zero singular value means the data cannot see that direction at all. The ratio $s_{\max}/s_{\min}$ for nonzero singular values is the spectral **condition number on the identifiable subspace**. A rank-deficient map has no inverse on the full input space; its full inverse condition number is infinite rather than this finite restricted ratio. A large condition number warns of sensitivity; it does not by itself identify whether the cause is redundant coordinates or missing physical information.

## Probability from normalization to conditioning

For a discrete variable $X$, the probabilities satisfy $p(x)\geq0$ and $\sum_xp(x)=1$. For a continuous variable the density satisfies $p(x)\geq0$ and $\int p(x)\,dx=1$. A density has units inverse to its integration variable: if $x$ is a position, $p(x)$ has units of inverse length. Probabilities of intervals are dimensionless integrals. A density can exceed one without violating probability.

The expectation of an observable $f$ is

$$
\mathbb E_p[f]=\int f(x)p(x)\,dx,
$$ 

with sums replacing integrals for discrete variables. Its variance is $\operatorname{Var}(f)=\mathbb E[(f-\mathbb E f)^2]=\mathbb E[f^2]-(\mathbb E f)^2$. Covariance is $\operatorname{Cov}(f,g)=\mathbb E[(f-\mathbb E f)(g-\mathbb E g)]$.

A joint distribution $p(x,y)$ gives the marginal $p(x)=\int p(x,y)dy$ and the conditional density $p(y\mid x)=p(x,y)/p(x)$ wherever $p(x)>0$. Therefore $p(x,y)=p(y\mid x)p(x)$. Writing this factorization in the opposite order yields Bayes' rule,

$$
p(x\mid y)=\frac{p(y\mid x)p(x)}{p(y)},\qquad
p(y)=\int p(y\mid x)p(x)\,dx.
$$ 

In inference, the prior $p(x)$ describes information before observing $y$, the likelihood $p(y\mid x)$ is the measurement model, and the posterior $p(x\mid y)$ is the updated distribution. The denominator is the evidence or normalization. Independence means $p(x,y)=p(x)p(y)$; zero covariance alone is weaker. For example, if $X$ is symmetric about zero and $Y=X^2$, then $\operatorname{Cov}(X,Y)=0$ but $Y$ is completely determined by $X$.

## The Gaussian distribution and a physical example

The normalized one-dimensional Gaussian is

$$
p(x)=\frac{1}{\sqrt{2\pi\sigma^2}}
\exp\left[-\frac{(x-\mu)^2}{2\sigma^2}\right].
$$

For completeness, set $I=\int_{-\infty}^{\infty}e^{-z^2/2}dz$. Squaring and using polar coordinates gives $I^2=\int_0^{2\pi}d\phi\int_0^\infty r e^{-r^2/2}dr=2\pi$. Thus $I=\sqrt{2\pi}$. A shift and rescaling normalize the stated density. Symmetry gives $\mathbb E[x]=\mu$; integration by parts gives $\mathbb E[(x-\mu)^2]=\sigma^2$.

For $d$ variables with positive-definite covariance $C$,

$$
p(x)=\frac{\exp[-(x-\mu)^TC^{-1}(x-\mu)/2]}
{(2\pi)^{d/2}\sqrt{\det C}}.
$$ 

Diagonalize $C=U\Lambda U^T$ and set $z=\Lambda^{-1/2}U^T(x-\mu)$. The exponent becomes $-\sum_a z_a^2/2$ and the volume element is $dx=\sqrt{\det C}\,dz$. This derives both the normalization and the interpretation of covariance eigenvectors as principal fluctuation directions. This rescaling is called **whitening**.

For a classical harmonic degree of freedom with energy $V(x)=kx^2/2$ in contact with a heat bath, the equilibrium density is proportional to $e^{-V/(k_BT)}$. Comparison with the Gaussian gives $\langle x^2\rangle=k_BT/k$. This is a concrete reason that quadratic losses, Gaussian uncertainty, and harmonic fluctuations repeatedly appear together. Nonquadratic energies or non-Gaussian detector noise require different models.

## Transforming a density

Let $x=f(z)$ be an invertible differentiable transformation. Conservation of probability means $p_X(x)dx=p_Z(z)dz$. In one dimension,

$$
p_X(x)=p_Z(f^{-1}(x))\left|\frac{d f^{-1}}{dx}\right|.
$$

In multiple dimensions the local volume factor is the absolute determinant of the Jacobian $J_{ab}=\partial f_a/\partial z_b$:

$$
p_X(f(z))=\frac{p_Z(z)}{|\det J(z)|}.
$$ 

The determinant appears because it is the volume of the transformed unit parallelepiped. Taking an absolute value handles orientation reversal. If $f$ is many-to-one, contributions from all preimages must be summed; the single-inverse formula no longer applies.

As an example, if $Z$ is a standard Gaussian and $X=e^Z$, then $z=\log x$, $dz/dx=1/x$, and $p_X(x)=e^{-(\log x)^2/2}/(x\sqrt{2\pi})$ for $x>0$. Omitting $1/x$ changes the distribution. Chapter 12 uses this same bookkeeping to build generative flows.

## Monte Carlo estimates and effective information

For independent samples $X_i\sim p$, the sample mean $\widehat f=N^{-1}\sum_i f(X_i)$ is unbiased because expectation is linear. Independence makes cross-covariances vanish, so

$$
\operatorname{Var}(\widehat f)=\frac{\operatorname{Var}_p(f)}{N}.
$$ 

The standard error decreases as $N^{-1/2}$, independent of dimension in this formal expression. Dimension still affects how hard it is to obtain samples and how large the observable variance becomes. For independent finite-variance samples, the central limit theorem gives an asymptotically Gaussian sampling distribution; a tiny sample or a heavy-tailed estimator may not justify Gaussian error bars.

For a stationary correlated sequence with $C(t)=\operatorname{Cov}(f_i,f_{i+t})$, expand the variance of the sum before simplifying:

$$
\operatorname{Var}(\widehat f)=\frac{1}{N^2}
\left[NC(0)+2\sum_{t=1}^{N-1}(N-t)C(t)\right].
$$

If correlations decay sufficiently quickly and $N$ is much larger than their range, define $\tau_{\mathrm{int}}=1/2+\sum_{t\geq1}C(t)/C(0)$. Then $\operatorname{Var}(\widehat f)\simeq 2\tau_{\mathrm{int}}C(0)/N$ and $N_{\mathrm{eff}}=N/(2\tau_{\mathrm{int}})$. The convention includes the factor $1/2$; other sources sometimes define a quantity twice as large. This estimate also assumes stationarity: it does not correct a chain trapped in one unseen region.

```python
import numpy as np
rng = np.random.default_rng(20261021)
# Independent thermal positions, with k_B T = 1 and k = 2.
x = rng.normal(0.0, np.sqrt(0.5), size=100_000)
estimate = np.mean(x*x)
stderr = np.std(x*x, ddof=1) / np.sqrt(x.size)
assert abs(estimate - 0.5) < 6 * stderr
print(f"<x^2> = {estimate:.5f} +/- {stderr:.5f}; exact 0.5")
```

For a Gaussian, $\mathbb E[x^4]=3\sigma^4$, obtained by another integration by parts. Thus $\operatorname{Var}(x^2)=2\sigma^4$ and the theoretical standard error here is $\sqrt{2}(0.5)/\sqrt N$. This supplies an analytic check on the numerical uncertainty estimate.

## Entropy, information, and distinguishability

For a discrete distribution the entropy is $H(p)=-\sum_xp(x)\log p(x)$, with $0\log0=0$. Natural logarithms measure information in nats. A uniform distribution over $K$ states has entropy $\log K$; a certain outcome has entropy zero. Entropy quantifies the spread of probability, not an intrinsic measure of scientific importance.

Relative entropy, or Kullback–Leibler divergence, is

$$
D_{\mathrm{KL}}(p\|q)=\sum_xp(x)\log\frac{p(x)}{q(x)}.
$$ 

It is infinite if $q$ vanishes where $p$ is positive. To show nonnegativity when support is adequate, use $\log u\leq u-1$ with $u=q/p$:
$-\sum p\log(q/p)\geq\sum p(1-q/p)=0$. If $q$ has mass outside the support of $p$, the same argument gives a nonnegative lower bound after restricting the sum. Equality requires identical distributions up to probability-zero sets. KL is not symmetric and is not a distance metric.

The cross-entropy is $H(p,q)=-\sum p\log q=H(p)+D_{\mathrm{KL}}(p\|q)$. When $p$ is fixed, minimizing cross-entropy minimizes KL. For continuous densities differential entropy depends on coordinates and reference units, but KL between two densities transformed consistently does not: their Jacobian factors cancel in the ratio and the integration measure transforms accordingly.

Mutual information is $I(X;Y)=D_{\mathrm{KL}}(p(x,y)\|p(x)p(y))$. It measures statistical dependence and is zero exactly under independence. It does not tell us which variable causes the other.

## Taylor expansions, gradients, and error budgets

For a smooth scalar function of many real parameters,

$$
L(\theta+\delta)=L(\theta)+g^T\delta
+\frac12\delta^TH\delta+O(\|\delta\|^3),
$$ 

where $g_a=\partial L/\partial\theta_a$ and $H_{ab}=\partial_a\partial_bL$. The gradient gives first-order change; the Hessian gives local curvature. At a stationary point $g=0$, positive curvature in every direction gives a strict local minimum. A zero eigenvalue needs higher-order analysis.

The central finite difference $[f(x+h)-f(x-h)]/(2h)$ has error $h^2f'''(x)/6+O(h^4)$, found by subtracting the two Taylor expansions. Floating-point subtraction also introduces error roughly proportional to machine precision divided by $h$. Making $h$ arbitrarily small eventually worsens the result. Automatic differentiation, introduced in Chapter 5, avoids this finite-difference truncation but still differentiates the implemented numerical function, which may itself approximate the intended physics.

Keep at least four errors conceptually distinct: model approximation, finite data or sampling, optimization, and numerical discretization. They need not add linearly, and a single validation score may mix them. A useful experiment varies one at a time: increase model capacity at fixed data; increase data at fixed capacity; improve optimization at fixed objective; and refine the solver grid at fixed physical model.

## Exercises and worked answers

1. A matrix has singular values $10$ and $10^{-3}$. What happens to output noise of size $10^{-2}$ along its second left singular vector when inverted?
2. Normalize $p(x)=c e^{-a x}$ on $x\geq0$, with $a>0$, and find its mean.
3. For stationary $C(t)=C(0)\rho^t$ with $0<\rho<1$, derive $N_{\mathrm{eff}}$ at large $N$.
4. Show that $D_{\mathrm{KL}}(p\|q)$ between two one-dimensional Gaussians of equal variance $\sigma^2$ is $(\mu_p-\mu_q)^2/(2\sigma^2)$.

**Worked answers.** (1) Noise becomes $10$ along the corresponding input vector, illustrating ill-conditioning. (2) Integration gives $1=c/a$, so $c=a$. Integration by parts gives $\mathbb E[x]=1/a$. (3) Summing a geometric series gives $2\tau_{\mathrm{int}}=1+2\rho/(1-\rho)=(1+\rho)/(1-\rho)$, so $N_{\mathrm{eff}}\simeq N(1-\rho)/(1+\rho)$. (4) Expand the log-density ratio as $[(x-\mu_q)^2-(x-\mu_p)^2]/(2\sigma^2)$ and average under $p$. The variance terms cancel; the remaining squared mean displacement gives the result.
